# Cálculo de Propiedades de una Mezcla Saturada Líquido + Vapor
# Determinar las propiedades termodinámicas de un sistema compuesto por una mezcla de 4 kg de líquido saturado y 2 kg de vapor saturado a T=130°C.

# Propiedades de la mezcla:

#$$V = V_f + X (V_g-V_f)$$
#$$H = h_f + X (h_g-h_f) = h_f + Xh_{fg} $$
#$$U = u_f + X (u_g-u_f) = u_f + Xu_{fg} $$
#$$S = s_f + X (s_g-s_f) = s_f + Xs_{fg} $$

image.png

```python
# Propiedades del sistema a 130°C
vf= 0.00107 #m3/kg
vg= 0.66808 #m3/kg
vfg = vg-vf # = 0.66701 m3/kg

uf= 546.1 #KJ/kg
ufg= 1993.4 #KJ/kg

hf = 546.38 #KJ/kg
hfg = 2173.7 #KJ/kg

Sf = 1.6346 #KJ/kgK
Sfg = 5.3919 # KJ/kg

X=1/3
```

```python
def M(Mf,Mfg,X):
  return Mf+X*Mfg
```

```python
print(f"Volumen Específico: {round(M(vf,vfg,X),4)} m3/kg")
```

```python
M(hf,hfg,X) #kJ/kg
```

```python
M(uf,ufg,X) #kJ/kg
```

```python
M(Sf,Sfg,X) # kJ/kgK
```

# Entalpía de una mezcla L+V



```python
# Vector:
import numpy
cal = numpy.linspace(start=0, stop=1, num=100)
```

```python
cal
```

```python
vectorH = M(hf,hfg,cal)
```

```python
vectorH
```

```python
import matplotlib.pyplot as plt
plt.plot(cal,vectorH)
plt.title('Entalpía de Mezcla ($Kj/kg$)')
plt.xlabel('Calidad')
plt.ylabel('Entalpía Promedio ($Kj/kg$)')
plt.grid(True)
plt.show()
```
